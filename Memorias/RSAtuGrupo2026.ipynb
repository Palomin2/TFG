```python
# PRELIMINARES
# Funciones que transforman un mismo dato de un tipo a otro. Algunas las hemos visto en AES

# Transformar un texto (entrecomillado) en un número en binario, 
# conforme al código ASCII (cada carácter, 8 bits), y al revés

def tex2bin(xtexto):
    return str(BinaryStrings().encoding(xtexto))

import sage.crypto.util

def bin2tex(xbin):
    cerosIzda=('0'*(8-len(xbin)%8))
    return sage.crypto.util.bin_to_ascii(cerosIzda+xbin)

# De binario a lista en binario, y al revés. Datos en decimal y en hexadecimal

def bin2lbin(xbin):
    return list(xbin)

def lbin2bin(xlista):
    xbin = '' 
    for ii in xlista:
        xbin = xbin + str(ii)
    return(xbin) 

def tex2lbin(xtexto):
    return bin2lbin(tex2bin(xtexto))

def dec2bin(xnum):
    return xnum.str(2)

def bin2dec(xbin): 
    return(Integer(xbin,2))

def dec2hex(xdec): 
    return(xdec.str(16)) 

def hex2dec(xhex): 
    return(Integer(xhex,16))
```
